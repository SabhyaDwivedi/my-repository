#### Chart - 1

````python
#create a scatter plot of room_type and price
sns.scatterplot(x='room_type',y='price',data=air_df)

#add labels
plt.xlabel('Room Types')
plt.ylabel('Price')
plt.title('Scatter Plot of Price by Room Type')

plt.show()
````

#### Chart - 2

````python
# Chart - 2 visualization code
#define colors
palette = {'Entire home/apt':'skyblue','Private room':'red','Shared room':'lightgreen'}


#create a box plot of minimum nights
sns.boxplot( x='room_type',y='minimum_nights', data=air_df,palette=palette)

# Set the y-axis limits focuses on the relevant range of minimum nights values.
plt.ylim(0, 30)

# Add labels and title
plt.xlabel('Room Type')
plt.ylabel('Minimum Nights')
plt.title('Distribution of Minimum Nights by Room Type')

# Display the plot
plt.show()

````

#### Chart - 3

````python
# Chart - 3 visualization code
#group by room type and calculate the average of reviews per month
mean_review_per_month = air_df.groupby('room_type')['reviews_per_month'].mean().reset_index() #used reset_index() converted mean_review_per month(series beacause of groupby) into dataframe

#create a bar plot
plt.figure(figsize=(8,6))
sns.barplot(x='room_type', y='reviews_per_month', data = mean_review_per_month, palette = 'Set1')

#add labels
plt.xlabel('Room Type')
plt.ylabel('Average Reviews per Month')
plt.title('Average Reviews per Month by Room Type')

#show
plt.show()

````

#### Chart - 4

````python
# Chart - 4 visualization code
#create bar/count plot for distribution of room types
plt.figure(figsize=(8,6))
ax=sns.countplot(x='room_type',data = air_df, palette ='Set2' )

# #iterating to find exact count using axis ax object
for bars in ax.containers:
    ax.bar_label(bars)

#add labels and title
plt.xlabel('Room Type', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Distribution of Room Types', fontsize=14)

#show()
plt.show()
````

#### Chart - 5

````python
# Chart - 5 visualization code
#create count/bar plot for listings across neighbourhood groups
plt.figure(figsize=(10, 6))
ax=sns.countplot(x='neighbourhood_group', hue='room_type', data=air_df)

#iterating to find exact count using axis ax object
for bars in ax.containers:
    ax.bar_label(bars)

#add title and labels
plt.title('Count of Listings by Neighbourhood Group and Room Type')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Count of Listings')

#add legend and show
plt.legend(title='Room Type')
plt.tight_layout()
plt.show()
````

#### Chart - 6

````python
# Chart - 6 visualization code
#group by neighbourhood groups and number of reviews colums
popular_location = air_df.groupby('neighbourhood_group')['number_of_reviews'].sum().reset_index()
print(popular_location) #this will give us count for each neighbourhood group

#create a bar plot
plt.figure(figsize=(8,6))
sns.barplot(x='neighbourhood_group', y='number_of_reviews', data = popular_location, palette='viridis')

#add labels
plt.xlabel('Neighbourhood Groups')
plt.ylabel('Number of Reviews')
plt.title('Total Number of Reviews by Neighbourhood Group')

#show
plt.show()
````

#### Chart - 7

````python
# Create the box plot
plt.figure(figsize=(10, 6))
sns.boxplot(x='neighbourhood_group', y='availability_365', data=air_df, palette='Set1')

# Add title and labels
plt.title('Distribution of Availability by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Availability (Days)')

# Show the plot
plt.tight_layout()
plt.show()
````

#### Chart - 8

````python
# Chart - 8 visualization code
#Group by neighbourhood group and sum the price then sort them using sort_values()
highest_price = air_df.groupby('neighbourhood_group',as_index= False)['price'].sum().sort_values('price', ascending=False)

#create a bar plot
plt.figure(figsize=(8,6))
sns.barplot(x='neighbourhood_group', y='price', data = highest_price, palette='Set2')

#add labels
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price')
plt.title('Total Price by Neighbourhood Group')

#show
plt.show()
````

#### Chart - 9

````python
# Chart - 9 visualization code
# Group the data by 'room_type' and calculate the average price for each room type
prefference= air_df.groupby('room_type', as_index=False)['number_of_reviews'].sum()

#create a bar plot
plt.figure(figsize=(8,6))
sns.barplot(x='room_type',y='number_of_reviews',data = prefference, palette='Set1')

#add labels
plt.xlabel('Room Type')
plt.ylabel('Number of Reviews')
plt.title('Total Number of Reviews for Airbnb Listings by Room Type')

#show
plt.show()
````

#### Chart - 10

````python
# Chart - 10 visualization code
# Scatter plot: Price vs Number of reviews
plt.figure(figsize=(10, 6))
plt.scatter(air_df['price'], air_df['number_of_reviews'], alpha=0.5)

#add labels and title
plt.title('Price vs Number of Reviews')
plt.xlabel('Price')
plt.ylabel('Number of Reviews')

#add grid and show
plt.grid(True)
plt.show()
````

````python
correlation_coefficient = air_df['price'].corr(air_df['number_of_reviews'])

print("Correlation coefficient between price and number of reviews:", correlation_coefficient)
````

#### Chart - 11

````python
# Chart - 11 visualization code
#group host names' and number of reviews
hosts= air_df.groupby('host_name')['number_of_reviews'].sum().reset_index()

#sort the list for to get host with most reviews using sort_values and head()
sorted_hosts= hosts.sort_values('number_of_reviews', ascending = False)
top_hosts= sorted_hosts.head(10)

print(top_hosts)
````

````python
#create a bar plot
plt.figure(figsize=(10,8))
sns.barplot(x='host_name',y='number_of_reviews',data = top_hosts)

#add labels and show
plt.xlabel('Host Name')
plt.ylabel('Number of Reviews')
plt.title('Top hosts by Most Number of Reviews')
plt.tight_layout()
plt.xticks(rotation=45, ha='right')
plt.show()
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
#select numeric data types
numeric_columns = air_df.select_dtypes(include=['int64','float64'])

#create correlation matrix
correlation = numeric_columns.corr()

#plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(correlation, annot = True, linewidths= 0.5, linecolor = 'Black', fmt='.2f', cmap= 'coolwarm')

#add title and show
plt.title(' Correlation Heatmap')
plt.show()
````
